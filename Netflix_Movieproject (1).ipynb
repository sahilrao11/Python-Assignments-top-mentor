**categorizing Vote_Average column**
We would cut the Vote_Average values and make 4 categories: popular average
below_avg not_popular to describe it more using catigorize_col() function
provided above.

```python
def catigorize_col (df, col, labels):
    edges = [df[col].describe()['min'],
             df[col].describe()['25%'],
             df[col].describe()['50%'],
             df[col].describe()['75%'],
             df[col].describe()['max']]
    df[col] = pd.cut(df[col], edges, labels = labels, duplicates='drop')
    return    
```

```python
# define labels for edges
labels = ['not_popular', 'below_avg', 'average', 'popular']
# categorize column based on labels and edges
catigorize_col(df, 'Vote_Average', labels)
# confirming changes
df['Vote_Average'].unique()
```

# Data Visualization

# Q1: What is the most frequent genre in the dataset?

```python
df['Genre'].describe()
```

```python
import matplotlib.pyplot as plt

# Count the number of movies in each genre
genre_counts = df['Genre'].value_counts()

# Plot a bar chart
plt.figure(figsize=(10,6))
genre_counts.plot(kind='barh', color='green')  # horizontal bar chart
plt.xlabel('Number of Movies')
plt.ylabel('Genre')
plt.title('Genre Distribution')
plt.gca().invert_yaxis() 
plt.show()
```

```python
df['Vote_Average'].describe()
```

```python
import matplotlib.pyplot as plt

# Count how many movies have each vote average
vote_counts = df['Vote_Average'].value_counts().sort_index()  # sort by vote value

# Plot horizontal bar chart
plt.figure(figsize=(12,6))
vote_counts.plot(kind='barh', color='red')
plt.xlabel('Vote Average')
plt.ylabel('Number of Movies')
plt.title('Vote Average Distribution')
plt.show()
```
