```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage[missing_percentage > 0].sort_values(ascending=False)
```

```python
df.duplicated().sum()
```

```python
df[["Name", "Age", "OVA", "Nationality", "Club", "Position"]].head()
```

```python
df["Club"] = df["Club"].fillna("Unknown")
df["Position"] = df["Position"].fillna("Unknown")
df["Nationality"] = df["Nationality"].fillna("Unknown")
```

```python
print("Average Age:", df["Age"].mean())
```

```python
print("Average Overall Rating:", df["OVA"].mean())
```

```python
print("Youngest player age:", df["Age"].min())
print("Oldest player age:", df["Age"].max())
```

```python
df[["Name", "Age", "OVA", "Club", "Nationality"]].sort_values(
    by="OVA", ascending=False
).head(10)
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Age"], bins=20)

plt.title("Distribution of Player Ages")
plt.xlabel("Age")
plt.ylabel("Number of Players")

plt.show()
```

```python
top_nationalities = df["Nationality"].value_counts().head(10)

top_nationalities
```

```python
plt.figure(figsize=(10, 6))

top_nationalities.plot(kind="bar")

plt.title("Top 10 Player Nationalities")
plt.xlabel("Nationality")
plt.ylabel("Number of Players")

plt.xticks(rotation=45)

plt.show()
```

```python
df[["Age", "OVA"]].corr()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    df[["Age", "OVA"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Age and Overall Rating")

plt.show()
```
