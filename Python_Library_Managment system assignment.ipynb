```python
library = []
```

```python
library = [
    {
        "Book ID": "101",
        "Title": "Python",
        "Author": "ABC",
        "Status": "Available",
        "Student": ""
    }
]
```

```python
def add_book():

    book_id = input("Enter Book ID: ")

    for book in library:
        if book["Book ID"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "Book ID": book_id,
        "Title": title,
        "Author": author,
        "Status": "Available",
        "Student": ""
    }

    library.append(book)

    print("Book Added Successfully!")
```

```python
def display_books():

    if len(library) == 0:
        print("No Books Available")
        return

    for book in library:
        print(book)
```

```python
def search_book():

    print("1. Search by ID")
    print("2. Search by Title")
    print("3. Search by Author")

    choice = input("Enter Choice: ")

    if choice == "1":
        book_id = input("Enter Book ID: ")

        for book in library:
            if book["Book ID"] == book_id:
                print(book)
                return

    elif choice == "2":
        title = input("Enter Title: ")

        for book in library:
            if book["Title"].lower() == title.lower():
                print(book)
                return

    elif choice == "3":
        author = input("Enter Author: ")

        for book in library:
            if book["Author"].lower() == author.lower():
                print(book)
                return

    print("Book Not Found")
```

```python
def update_book():

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["Book ID"] == book_id:

            title = input("Enter New Title: ")
            author = input("Enter New Author: ")

            if title != "":
                book["Title"] = title

            if author != "":
                book["Author"] = author

            print("Book Updated")
            return

    print("Book Not Found")
```

```python
def delete_book():

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["Book ID"] == book_id:

            library.remove(book)

            print("Book Deleted")

            return

    print("Book Not Found")
```

```python
def issue_book():

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["Book ID"] == book_id:

            if book["Status"] == "Issued":
                print("Already Issued")
                return

            student = input("Enter Student Name: ")

            book["Status"] = "Issued"
            book["Student"] = student

            print("Book Issued Successfully")
            return

    print("Book Not Found")
```

```python
def return_book():

    book_id = input("Enter Book ID: ")

    for book in library:

        if book["Book ID"] == book_id:

            book["Status"] = "Available"
            book["Student"] = ""

            print("Book Returned")

            return

    print("Book Not Found")
```

```python
def available_books():

    for book in library:

        if book["Status"] == "Available":
            print(book)
```

```python
def issued_books():

    for book in library:

        if book["Status"] == "Issued":
            print(book)
```

```python
while True:

    print("\nLibrary Management System")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Update Book")
    print("5. Delete Book")
    print("6. Issue Book")
    print("7. Return Book")
    print("8. Available Books")
    print("9. Issued Books")
    print("10. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        update_book()

    elif choice == "5":
        delete_book()

    elif choice == "6":
        issue_book()

    elif choice == "7":
        return_book()

    elif choice == "8":
        available_books()

    elif choice == "9":
        issued_books()

    elif choice == "10":
        print("Thank You")
        break

    else:
        print("Invalid Choice")
```
